## 2. Four years of prices

The average price people paid each month across Zambia. The dot marks each food's highest price.

```python
fig, ax = plt.subplots(figsize=(12, 6.2))
for p in PRODUCTS:
    s = df[df["product"] == p].sort_values("month")
    ax.plot(s["month"], s["price_kwacha"], color=COLORS[p], lw=2.6, solid_capstyle="round")
    peak = s.loc[s["price_kwacha"].idxmax()]
    ax.scatter(peak["month"], peak["price_kwacha"], s=70, color=COLORS[p], edgecolor=BG, linewidth=1.5, zorder=4)
    ax.annotate(f"Peak K{peak['price_kwacha']:.0f}  ({peak['month']:%b %Y})", (peak["month"], peak["price_kwacha"]),
                xytext=(10, 9), textcoords="offset points", fontsize=9.5, color=COLORS[p], fontweight="bold",
                bbox=dict(boxstyle="round,pad=0.2", fc=BG, ec="none", alpha=0.9))
    last = s.iloc[-1]
    ax.annotate(f"{p}\nK{last['price_kwacha']:.0f}", (last["month"], last["price_kwacha"]), xytext=(10, 0),
                textcoords="offset points", va="center", fontsize=10, color=COLORS[p], fontweight="bold")
ax.yaxis.set_major_formatter(KWACHA)
ax.set_ylim(0, df["price_kwacha"].max() * 1.12)
ax.set_xlim(df["month"].min() - pd.Timedelta(days=20), df["month"].max() + pd.Timedelta(days=20))
date_axis(ax)
headline(fig, "Staple food prices peaked in early 2025, then fell sharply",
         "National average retail price, December 2022 to October 2026", right=0.84)
save(fig, "01_prices")
plt.show()

summary = (df.sort_values("month").groupby("product")
             .agg(first=("price_kwacha", "first"), peak=("price_kwacha", "max"), latest=("price_kwacha", "last")))
summary["change from peak"] = summary["latest"] / summary["peak"] - 1
summary.style.format({"first": "K{:.2f}", "peak": "K{:.2f}", "latest": "K{:.2f}", "change from peak": "{:+.1%}"})
```

**What I see:** All three foods rose from late 2022 and reached their highest prices in early 2025. Breakfast mealie meal peaked at **K364 in March 2025**, roller mealie meal at **K307** and maize grain at **K229**, both in February 2025. After the 2025 harvest, prices dropped quickly.

## 3. Then, at the peak, and now

A simple way to see the whole story: what each food cost at the start, at its most expensive, and today.

```python
labels = ["December 2022", "Highest price", "October 2026"]
shades = {"December 2022": "#4E5551", "Highest price": "#F07B3F", "October 2026": "#3DBE8B"}
rows = []
for p in PRODUCTS:
    s = df[df["product"] == p].sort_values("month")
    peak = s.loc[s["price_kwacha"].idxmax()]
    rows.append({"product": p, "December 2022": s["price_kwacha"].iloc[0], "Highest price": peak["price_kwacha"],
                 "peak month": f"{peak['month']:%b %Y}", "October 2026": s["price_kwacha"].iloc[-1]})
tn = pd.DataFrame(rows).set_index("product")

fig, ax = plt.subplots(figsize=(12, 6.2))
width = 0.26
for j, lab in enumerate(labels):
    xs = [i + (j - 1) * width for i in range(len(tn))]
    bars = ax.bar(xs, tn[lab], width=width * 0.92, color=shades[lab], label=lab, zorder=2)
    for x, v, (prod, r) in zip(xs, tn[lab], tn.iterrows()):
        txt = f"K{v:.0f}" + (f"\n{r['peak month']}" if lab == "Highest price" else "")
        ax.text(x, v + 6, txt, ha="center", va="bottom", fontsize=9.5, color=INK, fontweight="bold" if lab != "December 2022" else "normal")
ax.set_xticks(range(len(tn)))
ax.set_xticklabels([f"{p}\n({df[df['product'] == p]['unit'].iloc[0]})" for p in tn.index], fontsize=11, color=INK)
ax.tick_params(axis="x", length=0)
ax.yaxis.set_major_formatter(KWACHA)
ax.set_ylim(0, tn["Highest price"].max() * 1.22)
ax.legend(frameon=False, ncol=3, loc="upper right", fontsize=10)
headline(fig, "Prices more than doubled, then fell back to around mid-2023 levels",
         "Price at the start, at the highest point and today, for each food")
save(fig, "02_then_peak_now")
plt.show()

tn["rise to peak"] = tn["Highest price"] / tn["December 2022"] - 1
tn["fall since peak"] = tn["October 2026"] / tn["Highest price"] - 1
tn["today vs Dec 2022"] = tn["October 2026"] / tn["December 2022"] - 1
tn[["December 2022", "Highest price", "peak month", "October 2026", "rise to peak", "fall since peak", "today vs Dec 2022"]].style.format(
    {"December 2022": "K{:.2f}", "Highest price": "K{:.2f}", "October 2026": "K{:.2f}",
     "rise to peak": "{:+.0%}", "fall since peak": "{:+.0%}", "today vs Dec 2022": "{:+.0%}"})
```

**What I see:** Between December 2022 and early 2025, prices **more than doubled**. Breakfast mealie meal went from K172 to K364, and maize grain rose even faster, from K88 to K229. Since then, prices have fallen by **35% to 53%**. A 25 kg bag of breakfast mealie meal now costs **K236**, about the same as in mid-2023. Prices are still higher than in December 2022, though: by 37% for breakfast mealie meal, 31% for roller mealie meal and 22% for maize grain.

The average price across each year's months. 2026 only includes January to October.

**What I see:** Even though prices hit their highest point in early 2025, **2024 was the most expensive year overall**. Prices stayed high for almost the whole of 2024, while in 2025 they fell sharply after the harvest. For example, a bag of breakfast mealie meal cost **K328 on average in 2024**, compared with K309 in 2025 and K273 so far in 2026.

## 5. The yearly cycle

Each line is one year, so you can compare the same months across years. The shaded area marks January to March, the months before the harvest.

```python
year_style = {2023: ("#4E5551", 1.8), 2024: ("#F07B3F", 2.0), 2025: (INK, 2.6), 2026: ("#3DBE8B", 3.0)}
fig, axes = plt.subplots(1, 3, figsize=(14, 5.6), sharex=True)
for ax, prod in zip(axes, PRODUCTS):
    t = df[df["product"] == prod].copy()
    t["year"], t["m"] = t["month"].dt.year, t["month"].dt.month
    ends = []
    for y, (c, lw) in year_style.items():
        ty = t[t["year"] == y]
        ax.plot(ty["m"], ty["price_kwacha"], color=c, lw=lw, solid_capstyle="round")
        ends.append([ty["price_kwacha"].iloc[-1], ty["m"].iloc[-1], y, c])
    # Spread the year labels so they never overlap
    gap = (t["price_kwacha"].max() - t["price_kwacha"].min()) * 0.065
    ends.sort()
    for i in range(1, len(ends)):
        ends[i][0] = max(ends[i][0], ends[i - 1][0] + gap)
    for yv, m, y, c in ends:
        label_c = "#7D857F" if y == 2023 else c
        ax.text(m + 0.25, yv, str(y), color=label_c, fontsize=9.5, fontweight="bold", va="center")
    ax.axvspan(0.5, 3.5, color="#F07B3F", alpha=0.06, lw=0)
    ax.set_title(prod, loc="left", fontsize=12, fontweight="bold", color=COLORS[prod])
    ax.set_xticks(range(1, 13))
    ax.set_xticklabels(["J", "F", "M", "A", "M", "J", "J", "A", "S", "O", "N", "D"])
    ax.set_xlim(0.5, 13.3)
    ax.yaxis.set_major_formatter(KWACHA)
axes[0].text(2, axes[0].get_ylim()[0] + 4, "Jan–Mar", ha="center", fontsize=8.5, color="#F07B3F")
headline(fig, "Since 2025, prices peak before the harvest and fall after it",
         "Price by month, one line per year. In 2023 and 2024, prices kept rising almost all year.",
         top=0.76, left=0.05, right=0.98, bottom=0.12)
save(fig, "04_yearly_cycle")
plt.show()

# In which month did each year's maize grain price peak?
mg = df[df["product"] == "Maize grain"].copy()
mg["year"] = mg["month"].dt.year
mg.loc[mg.groupby("year")["price_kwacha"].idxmax(), ["year", "month", "price_kwacha"]].style.format(
    {"month": "{:%b %Y}", "price_kwacha": "K{:.2f}"}).hide(axis="index")
```

**What I see:** In 2025 and 2026, prices followed a clear pattern: they were **highest in February or March**, the months before the harvest, then **lowest around August to October**, after the harvest. 2023 and 2024 looked different: mealie meal prices kept rising almost all year, and maize grain dipped only slightly after the harvest before rising again. 2025 (the black line) started higher than any other year, then fell the furthest.

*Note: December 2022 is the only month from 2022, so 2022 is not shown as its own line.*

## 6. Where you live matters

ZamStats publishes the cheapest and most expensive district for each food. We have 10 of these snapshots between December 2023 and October 2026. There is one chart for each food. The diamond shows the national average price for that month.

```python
def district_chart(p, fname):
    d = districts[districts["product"] == p].sort_values("month").reset_index(drop=True)
    nat = df[df["product"] == p].set_index("month")["price_kwacha"]
    d["national"] = d["month"].map(nat)
    unit = d["unit"].iloc[0]

    fig, ax = plt.subplots(figsize=(12, 7))
    y = list(range(len(d)))
    ax.hlines(y, d["min_price_kwacha"], d["max_price_kwacha"], color="#2E3430", lw=7, capstyle="round", zorder=1)
    ax.scatter(d["min_price_kwacha"], y, s=90, color="#3DBE8B", edgecolor=BG, linewidth=1.5, zorder=3)
    ax.scatter(d["max_price_kwacha"], y, s=90, color="#F07B3F", edgecolor=BG, linewidth=1.5, zorder=3)
    ax.scatter(d["national"], y, s=70, marker="D", color=INK, edgecolor=BG, linewidth=1.2, zorder=4)
    span = d["max_price_kwacha"].max() - d["min_price_kwacha"].min()
    pad = span * 0.03
    for i, r in d.iterrows():
        names = [n.strip() for n in str(r["cheapest_district"]).split(",")]
        cheapest = names[0] if len(names) < 3 else f"{names[0]} + {len(names) - 1} more"
        ax.text(r["min_price_kwacha"] - pad, i, cheapest, ha="right", va="center", fontsize=8.8, color="#3DBE8B")
        ax.text(r["max_price_kwacha"] + pad, i, f"{r['most_expensive_district']}  ({r['gap_ratio']:.1f}×)", ha="left", va="center", fontsize=8.8, color="#F07B3F")
    ax.set_yticks(y)
    ax.set_yticklabels(d["month"].dt.strftime("%b %Y"), fontsize=10, color=INK)
    ax.tick_params(axis="y", length=0)
    ax.grid(axis="y", visible=False)
    ax.grid(axis="x", color=GRID)
    ax.xaxis.set_major_formatter(KWACHA)
    ax.set_xlim(max(0, d["min_price_kwacha"].min() - span * 0.55), d["max_price_kwacha"].max() + span * 0.55)
    ax.invert_yaxis()
    fig.text(0.05, 0.835, "●", color="#3DBE8B", fontsize=12, va="center"); fig.text(0.068, 0.835, "Cheapest district", color=MUTED, fontsize=10, va="center")
    fig.text(0.22, 0.835, "◆", color=INK, fontsize=11, va="center");    fig.text(0.238, 0.835, "National average", color=MUTED, fontsize=10, va="center")
    fig.text(0.39, 0.835, "●", color="#F07B3F", fontsize=12, va="center"); fig.text(0.408, 0.835, "Most expensive district (× cheapest)", color=MUTED, fontsize=10, va="center")
    headline(fig, f"{p}: the most expensive district paid up to {d['gap_ratio'].max():.1f}× the cheapest",
             f"{unit}: cheapest and most expensive district in 10 months", top=0.80, left=0.16, text_left=0.05)
    save(fig, fname)
    plt.show()

    print(f"{p}: how often each district was the most expensive")
    print(d["most_expensive_district"].value_counts().to_string())
    print("Average gap:", round(d["gap_ratio"].mean(), 2), "times the cheapest\n")

district_chart("Breakfast mealie meal", "05a_districts_breakfast")
district_chart("Roller mealie meal", "05b_districts_roller")
district_chart("Maize grain", "05c_districts_maize")
```

**What I see:** For breakfast mealie meal, a few remote districts consistently pay the most. **Chilubi was the most expensive in 5 of the 10 snapshots** and **Chiengi in 3** (they shared the top spot once more). Their price stayed around **K380 to K420** even when the national average fell, so the gap grew: in October 2026, a bag in Chilubi cost **2.2 times** as much as in the cheapest district, the widest gap for breakfast mealie meal. One possible reason is the cost of transporting goods to remote areas, but this data alone can't prove that.

**Roller mealie meal** follows a similar pattern: Chilubi was the most expensive in 4 of the 10 snapshots, and the gap reached **2.3 times** the cheapest district. For **maize grain**, the most expensive place changes more often (Kazungula, Zambezi and Mambwe each appear twice), but the gap is the widest of all three foods, up to **2.7 times**.

*Note: the cheapest breakfast mealie meal price for January 2026 is printed as "268.5.00" in the bulletin. It was read as K268.50 and flagged in the `note` column.*

## 7. Key findings

1. **Prices more than doubled, then fell.** Between December 2022 and early 2025, breakfast mealie meal rose from K172 to K364, roller mealie meal from K139 to K307 and maize grain from K88 to K229. Since then, they have fallen by 35%, 41% and 53%.
2. **Prices are back to around mid-2023 levels.** A 25 kg bag of breakfast mealie meal costs K236 today, close to what it cost in mid-2023, but still 37% more than in December 2022.
3. **2024 was the most expensive year overall.** Prices peaked in 2025, but they stayed high for almost all of 2024, making it the most expensive year on average for all three foods.
4. **Where you live still matters.** Remote districts such as Chilubi and Chiengi were usually the most expensive for mealie meal. The most expensive district paid up to 2.2 times the cheapest price for breakfast mealie meal, 2.3 times for roller mealie meal and 2.7 times for maize grain.

## 8. Data and limitations

- **Source:** national average retail prices from 10 ZamStats Monthly Bulletins (December 2023 to October 2026 issues), covering every month from December 2022 to October 2026.
- **Checks:** months that appear in more than one bulletin were compared and matched exactly. 18 months come from a single bulletin (see the `bulletins_confirming` column).
- **Prices as paid:** these are the prices people paid at the time. They do not account for changes in incomes or in the cost of other goods.
- **National averages:** prices vary a lot between districts, so a national average does not describe every household.
- **District data:** only the cheapest and most expensive district are published, for 10 months, so this is not a full district comparison.
- **What this does not show:** the data shows *when* prices changed, not *why*. Explanations such as the drought or harvest timing are context, not conclusions proven by this data.
